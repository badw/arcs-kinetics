```python
from ase.io import read
from hofmann import StructureScene
import matplotlib.pyplot as plt 

traj = read("./notebooks/nanoreactor_forward/trajectory.traj", index=":")
scene = StructureScene.from_ase(traj)
len(traj)
#scene.render_animation("output.gif")
```

```python
scene.render_mpl()
scene.view.look_along([1,1,1])
scene.view.zoom = 0.6
plt.show()
```

```python
scene.render_animation(
    "output.mp4",
    fps=60,               # default: 30
    dpi=100,              # default: 150
    figsize=(6.0, 6.0),   # default: (5.0, 5.0)
)
```
