```python
from tinydb import TinyDB,Query 

db = TinyDB("db.json")
# block deleting/editing
def blocked(*a, **k): raise PermissionError("database is append-only")
db.remove = db.update = db.upsert = db.truncate = db.drop_tables = blocked

def add_reaction(reactants, products, **data):
    if not db.contains(Query().reactants == sorted(reactants)):
        db.insert({"reactants": sorted(reactants), "products": products, **data})

def compounds_list():
    return sorted({c for r in db for c in r["reactants"] + r["products"]})


add_reaction(["H2O", "SO3"], ["H2SO4"], Ea=1, dE=-0.1)
add_reaction(["H2O", "NO2"], ["HNO3", "HNO2"], Ea=0.8, dE=-0.2)
```

```python
db.all()
```
